```python
import matplotlib.pyplot as plt

metrics = ((0.86, 24.94), (0.87, 25.59), (0.89, 26.43), (0.90, 27.11), (0.91, 27.90))
view_counts = [1, 2, 3, 6, 9]


# Separate SSIM and PSNR metrics
ssim_values = [metric[0] for metric in metrics]
psnr_values = [metric[1] for metric in metrics]

# Create the first plot for SSIM
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.plot(view_counts, ssim_values, marker='o', linestyle='-', color='cornflowerblue')
plt.xscale('log')
plt.ylim(0.85, 1)
plt.title('SSIM w.r.t. View Count')
plt.xlabel('View Counts (log scale)')
plt.ylabel('SSIM')
plt.xticks(view_counts, labels=view_counts)
plt.grid(True)

# Create the second plot for PSNR
plt.subplot(1, 2, 2)
plt.plot(view_counts, psnr_values, marker='o', linestyle='-', color='cornflowerblue')
plt.xscale('log')
plt.ylim(24, 30)
plt.title('PSNR w.r.t. View Count')
plt.xlabel('View Counts (log scale)')
plt.ylabel('PSNR')
plt.xticks(view_counts, labels=view_counts)
plt.grid(True)

# Display the plots
plt.tight_layout()
plt.savefig('metrics_plots.pdf', format='pdf')
plt.show()
```
